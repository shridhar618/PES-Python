```python
import csv
```

```python

with open('shridhar.csv','w')as f:
    writer=csv.writer(f)
    writer.writerow(["name","age","city"])
    writer.writerow(["shridhar","21","shiroor"])
    writer.writerow(["virat","37","delhi"])
```

```python
with open('example.csv','r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
with open('example.csv') as file:
    reader=csv.reader(file,skipinitialspace=True)
    for row in reader:
        print(row)
```

```python
with open('example.csv','r') as file:
    reader=csv.reader(file,delimiter=';')
    for row in reader:
        print(row)
```

```python
with open('example.csv','r') as file:
    csv_reader=csv.DictReader(file)
    
    for row in csv_reader:
        print(row)
```

```python
with open('example.csv','r') as file:
    reader=csv.DictReader(file,delimiter=";")

    for row in reader:
        print(row)
```

```python
with open('example.csv','r') as file:
    reader=csv.DictReader(file,quotechar='"')

    for row in reader:
        print(row)
```

```python
fieldnames=["Name","age","Profession"]

with open('example.csv','r') as file:
    reader=csv.DictReader(file,fieldnames=fieldnames)

    for row in reader:
        print(row)
```

```python
with open('emp_data.csv','r') as file:
    csv_reader=csv.reader(file)
    count=0
    for row in csv_reader:
       if count==0:
           print(f'Column names are : {", ".join(row)}')
       else:
           print(f'\t {row[0]} works in the {row[1]} department , was born in {row[2]}')
           count+=1
        
    print(f'Processed {count} lines')
```

```python
data=[
    ['Name','Age','City'],
    ['Shubham',24,'Patna'],
    ['Mohit',25,'Danapur'],
    ['Rajnish',20,'Bengaluru']
]

with open('output.csv','w',newline='') as file:
    writer=csv.writer(file)
    writer.writerows(data)

with open('output.csv','r') as file:
    reader=csv.reader(file)

    for row in reader:
        print(row)
```
