# XML - FILES


## Element tree

## import the package

```python
import xml.etree.ElementTree as xml
```

# Create the data

## Normally written as a list of dictionaries

```python
books=[
    {"name":"Pride & Prejudice",
     "Author":"Jane Austen",
     "Price":240,
     "Category":"fiction"},
    {"name":"The dairy of a young Girl",
     "Author":"Anne Frank",
     "Price":500,
     "Category":"Non-fiction"},
    {"name":"Five on the treasure island",
     "Author":"Enid Blyton",
     "Price":250,
     "Category":"fiction"}
]
     
```

# Create the root node

```python
root=xml.Element("Book_store")
#creating an object of element whose name is book_store
```

```python
type(root)
```

# iterate over the list & create a tree

```python
for b in books:
    b_element=xml.Element('Book')
    #Create on instance of Element with name Book
    root.append(b_element)
    #Attach the node to the root node
    b_element.set('Category',b['Category'])
    #Add attributes to the node
    #First parameter - name of the attribute
    #Second parameter - value of the attribute
    title=xml.SubElement(b_element, 'Name')
    #Create a child node
    #First parameter - parent node
    #Second parameter - name of the child
    title.text=b['Name']
    #Add the information to the node
    author=xml.SubElement(b_element, 'Author')
    author.text=b['Author']
    price= xml.SubElement(b_element, 'Price')
    price.text=str(b['Price'])
```

```python
type(b_element),type(b),type(title),type(author),type(price)
```

```python
tree=xml.ElementTree(root)
#Create an instance of Element Tree by passing the root node
```

```python
type(tree)
```

# Create an XML file

```python
with open('Books.xml', 'wb')as f:
    tree.write(f)
```

# Read from an XML file

```python
tree=xml.ElementTree(file='Books.xml') #get the tree structure from the file
```

```python
root=tree.getroot()#once the tree structure is found get the rootnode that already exists in the file
```

```python
Books=[]
#an emptty list to store the data from xml file
```

# Iterate over the nodes to get the data

```python
for b in root.findall('Book'):
    #iterates over all nodes that are named book
    b_data={}
    #create an empty dictionary to store each parent & its children value
    b_data['Ctaegory']=b.get('Category')
    #Create the first key-value pair
    #the index of the dictionary is represented by the key
    #the value of the pair is got from the node
    for d in b:
        b_data[d.tag]=d.text
        #get the child node's data
        #tag gets the name of the child node
        #text gets the value of the child node
    Books.append(b_data)
```

```python
print(Books)
```

# Modification of the existing XML file

## Get the tree

```python
tree=xml.ElementTree(file='Books.xml')
root=tree.getroot()
```

## Change the price of every book

```python
for p in root.iter("Price"):
    #traverse over the iterable object of price
    p_val=int(p.text)
    #Type conversion of string to list
    p_val+=15
    p.text=str(p_val)
    
```

## Write the changed data into the file

```python
with open('Books.xml','wb')as f:
    tree.write(f)
```
