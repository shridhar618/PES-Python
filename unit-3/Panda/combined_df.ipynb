# Combine data Frames

```python
import pandas as pd
```

```python
sales=pd.read_csv("sales.csv")
products=pd.read_csv("products.csv")
```

```python
#if the number of unique occurences matches the num of records of products , the data frames can be merged or combined
```

```python
len(sales)
```

```python
merge_inner=pd.merge(sales, products, 
                     on='ProductID', how='inner')
```

```python
type(merge_inner) #when u work on a data frame it gives u a dataframe
```

```python
len(merge_inner) 
```

```python
merge_inner
```

```python
merge_left=pd.merge(sales, products, 
                     on='ProductID', how='left') #left join
```

```python
len(merge_left)
```

```python
merge_left
```

```python
merge_right=pd.merge(sales, products, 
                     on='ProductID', how='right') #right join
```

```python
len(merge_right)
```
