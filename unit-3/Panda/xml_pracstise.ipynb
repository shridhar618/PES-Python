```python
import xml.etree.ElementTree as xml
```

```python
books=[
    {
        "Name":"Pride & Prejudice",
        "Author":"Jane Austen",
        "Price":250,
        "Category":"Fiction"
    },
    {
        "Name":"The Dairy of a young Girl",
        "Author":"Anne Frank",
        "Price":500,
        "Category":"Non-Fiction"
    },
    {
    "Name":"Five on the treasure island",
     "Author":"Enid Blyton",
     "Price":250,
     "Category":"fiction"
    }
]
```

```python
root=xml.Element("Book_store")
```

```python
for b in books:
    b_element=xml.Element('Books')

    root.append(b_element)

    b_element.set('Category',b['Category'])

    title=xml.SubElement(b_element,"Name")
    title.text=b["Name"]

    author=xml.SubElement(b_element,"Author")
    author.text=b["Author"]

    price=xml.SubElement(b_element,"Price")
    price.text=b["Price"]
    
```

```python
tree=xml.ElementTree(root)
```

```python
type(tree)
```

```python
with open("Books.xml",'wb') as f:
    tree.write(f)
```

These lines open an XML file and get its main (root) tag so you can read or modify the data inside it.

```python
tree=xml.ElementTree(file='Books.xml')
root=tree.getroot()
```

```python
books=[]
```

This line looks for all XML tags named <Books> inside the root element and loops over them one by one.     for b in root.findall("Books"):

```python
for b in root.findall("Books"):
    b_data={}

    b_data["Category"]=b.get("Category")
    for d in b:
        b_data[d.tag]=d.text

    books.append(b_data)
```

```python
for b in books:
    print(b)
```
