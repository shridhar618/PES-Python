```python
import xml.etree.ElementTree as xml
```

```python
books=[
    {
        "Name":"Pride & Prejudice",
        "Author":"Jane Austen",
        "Price":250,
        "Category":"Fiction"
    },
    {
        "Name":"The Dairy of a young Girl",
        "Author":"Anne Frank",
        "Price":500,
        "Category":"Non-Fiction"
    },
    {
    "Name":"Five on the treasure island",
     "Author":"Enid Blyton",
     "Price":250,
     "Category":"fiction"
    }
]
```

```python
root=xml.Element("Book_store")
```

```python
type(root)
```

```python
for b in books:
    b_element=xml.Element('Books')

    root.append(b_element)
    
    # <Book_store>
    #     <Book></Book>
    # </Book_store>


    b_element.set('Category',b['Category'])

    # <Book Category="fiction">

    title=xml.SubElement(b_element,"Name")
    title.text=b["Name"]

    # <Name>Pride & Prejudice</Name>

    author=xml.SubElement(b_element,"Author")
    author.text=b["Author"]

    price=xml.SubElement(b_element,"Price")
    price.text=str(b["Price"])


# <Book_store>
#     <Book Category="fiction">
#         <Name>Pride & Prejudice</Name>
#         <Author>Jane Austen</Author>
#         <Price>240</Price>
#     </Book>
# </Book_store>


    
```

```python
type(b_element),type(b),type(title),type(author),type(price)
```

```python
tree=xml.ElementTree(root)
```

```python
type(tree)
```

```python
with open("Books.xml",'wb')as f:
    tree.write(f)
```

```python
tree=xml.ElementTree(file='Books.xml')
root=tree.getroot()
```

```python
Books=[]
```

```python
for b in root.findall("Books"):

    b_data={}

    b_data["Category"]=b.get("Category")

    for d in b:
        b_data[d.tag]=d.text

    Books.append(b_data)
```

```python
for b in Books:
    print(b)
```

```python
tree=xml.ElementTree(file='Books.xml')
root=tree.getroot()
```

```python
for p in root.iter("Price"):
    #traverse over the iterable object of price
    p_val=int(p.text)
    #Type conversion of string to list
    p_val+=15
    p.text=str(p_val)
```
