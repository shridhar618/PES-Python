# Cleaning the data-Preprocessing

# clean the column names

```python
df.columns=df.columns.str.strip()
```

```python
df.columns
```

# clean the data in the column

```python
df['Amount']=(df['Amount'].astype(str)
              .str.replace('"','',regex=False)
              .str.replace(',','',regex=False)
              .str.extract(r'(\d+)',expand=False))
              
```

# Convert strings to  numeric

```python
df['Amount']=pd.to_numeric(df['Amount'],errors='coerce')
print(df.dtypes)
'''errors = 'coerce' makes all non numeric data as NAN'''
```

```python
df['Amount']
```

# Fill the missing data

# 1.median is used to fill 

```python
df['Amount']=df['Amount'].fillna(df['Amount'].median())
```

```python
df['Amount']
```

```python
df['Qty']
```

```python
df['Qty']=pd.to_numeric(df['Qty'],errors='coerce')
print(df.dtypes)
```

# 2. mean is used to replace NAN

```python
df['Qty']=df['Qty'].fillna(df['Qty'].mean())
df['Qty']
```

# 3. mode is also used to fill NAN

```python
df.head()
```

# Drop a column that has -: 1) NAN  2) Not needed for processing

```python
df=df.loc[:,~df.columns.str.contains("^Unnamed")] 
```

```python
df
```

# Equavalent options

```python
df=df.drop(columns=["Unnamed: 6"])
#give full name of the column
df=df.iloc[:,:-1]
#dropping by index position (but the problem is u can't drop in btw column), it works only for corner columns
df=df.dropna(axis=1, how="all")
#drop columns with all NAN or all missing values
```

```python
df.head()
```

# Standardize text columns

```python
df['Region']=df['Region'].str.strip().str.lower()
```

```python
df['Region']
```

# Removes quotes in all columns

```python
df=df.apply(lambda col: col.astype(str)
            .str.replace('"','',regex=False))
```

# Custom Aggregation 

```python
summary=df.groupby('Region').agg(
    Total=('Amount','sum'),
    Average_Amt=('Amount','mean'),
    Total_Tax=('Tax','sum'),
    Average_Tax=('Tax','mean'),
    Unique_Customers=('Customer','nunique')
)
    
```
