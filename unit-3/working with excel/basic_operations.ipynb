```python
print(ws.cell(row=2,column=3).value)
```

```python
print(ws.max_row)
print(ws.max_column)
```

```python
for row in range(2,ws.max_row+1):
    if ws.cell(row=row,column=1).value==102:
        ws.cell(row=row,column=3).value=95
        print("succes full work")
        break

wb.save("student.xlsx")
```

```python
for row in range(2,ws.max_row+1):
    if ws.cell(row=row,column=2).value=="Shubham":
        ws.cell(row=row,column=3).value=55
        print("Succes full")
        break

wb.save("student.xlsx")
```

Deleting the data

```python
# Delete a row

ws.delete_rows(3)
# Delete multiple row   delete 2 rows starting from row 2
ws.delete_rows(2,2)
```

```python
# Delete columns

ws.delete_cols(2)               # deletes column B

# Delete multiple columns       removes column B & C


ws.delete_cols(2,2) 
```
