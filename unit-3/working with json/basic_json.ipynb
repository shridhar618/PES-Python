```python

data=''' [
    {"Name":"Shubham","Age":23},
    {"Name":"Mohit","Age":25},
    {"Name":"Raman","Age":33}

]'''

student=json.loads(data)

print(student[0]["Name"],student[0]["Age"])
print(student[1]["Name"],student[1]["Age"])

for s in student:
    print(s["Name"],s["Age"])
```
